### Insight — `ano_modelo` and `zero_km`

In this dataset, `ano_modelo` and `zero_km` follow a deterministic relationship:

- All records where `zero_km = True` have `ano_modelo = null`.
- All records where `ano_modelo = null` correspond to `zero_km = True`.
- No non-zero-km vehicle has a null `ano_modelo`.

This indicates that the null values in `ano_modelo` are not random missing data, but instead represent an expected business rule for zero-km vehicles.

Therefore, `ano_modelo` should not be imputed without explicitly preserving the `zero_km` vehicle condition.

### 4.3 Cardinality and Domain Analysis

```python
df.nunique(dropna=False).sort_values()
```

```python
profile = pd.DataFrame({
    'dtype': df.dtypes.astype(str),
    'null_count': df.isna().sum(),
    'null_pct': df.isna().mean().mul(100).round(3),
    'unique_count': df.nunique(dropna=True),
    'unique_pct': (
        df.nunique(dropna=True) / df.notna().sum()
    ).mul(100).round(3)
})

profile.sort_values('unique_count')
```

```python
def classify_cardinality(row):
    n_unique = row['unique_count']

    if n_unique == 1:
        return 'constant'
    elif n_unique <= 10:
        return 'low'
    elif n_unique <= 500:
        return 'medium'
    else:
        return 'high'
```

```python
profile.sort_values('unique_pct')
```

### Cardinality Classification

For this analysis, columns will be classified primarily according to their `unique_count`:

- **Constant:** `unique_count = 1`
- **Low cardinality:** `unique_count <= 10`
- **Medium cardinality:** `10 < unique_count <= 500`
- **High cardinality:** `unique_count > 500`

`unique_pct` will be used as a complementary metric to understand the relative cardinality of each column.

These thresholds are project-specific analytical conventions adopted for this dataset and are intended to guide the appropriate depth of domain inspection. They should not be interpreted as universal cardinality standards.

### Constant Columns

`mes_referencia` and `ano_referencia` contain a single value in the current dataset.

This is expected because the file represents a single FIPE snapshot.

These columns require no further domain inspection at this stage, but they are relevant for historical datasets, where they become part of the snapshot reference and historical key.

### Insight — `codigo_fipe` and `nome_modelo`

`nome_modelo` and `codigo_fipe` have very similar cardinalities:

- `nome_modelo`: **11,391** unique values
- `codigo_fipe`: **11,396** unique values

The difference is explained by a small set of model names that are shared across different manufacturers.

The following five model names are associated with more than one FIPE code:

- `16-220 T 3-Eixos 2p (diesel)`
- `16-220 Turbo 2p (diesel)`
- `CR 125`
- `CR 250`
- `FENIX GOLD 240`

In these cases, the ambiguity is resolved by `nome_marca`. For example, the same `nome_modelo` can refer to different vehicles when associated with different manufacturers.

The following relationships were validated:

- Each `codigo_fipe` maps to exactly one `nome_marca` + `nome_modelo` combination.
- Each `nome_marca` + `nome_modelo` combination maps to exactly one `codigo_fipe`.
- `nome_modelo` alone is not globally unique across manufacturers.

Therefore, `codigo_fipe` behaves as a unique business identifier for the brand-model combination, while `nome_modelo` should not be used independently as a unique vehicle identifier.

This also confirms that the small difference in cardinality between `nome_modelo` and `codigo_fipe` is expected and does not indicate a data quality issue.
